Imports and Setup

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from skimage import data
from skimage.transform import resize
from skimage import io
```

Read and display image

```python
%matplotlib inline
```

```python
# Reading sample image from the skimage library
img = data.astronaut()  
print("Original Image Shape:", img.shape)
plt.imshow(img) #Reading an Image
plt.title("Original Image")
plt.axis("off")
plt.show() # This is needed to actually render the window, similar to plt.show()


# reading image from file using skimage io
# img_using_skimage_io = io.imread("practical1.png")
# plt.imshow(img_using_skimage_io)
# plt.axis("off")
# plt.show()


# # reading image from file using OpenCV
# img = cv2.imread("practical1.png")  # Loading the image using OpenCV but OpenCV loads images in BGR (Blue, Green, Red) format
# img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB) #Convert the image from BGR to RGB for correct color display
# plt.imshow(img)
# plt.title("Original Image")
# plt.axis("off")
# plt.show()
# print("Original Image Shape:", img.shape)
```

Resize a given image.

```python
resized_image = resize(img, (200, 200))  # stays in RGB format
plt.imshow(resized_image)
plt.title("Resized Image (200x200)")
plt.axis("off")
plt.show()
print("Resized Image Shape:", resized_image.shape)
```

Convert a given color image into a corresponding gray-scale image.

```python
# Convert to grayscale
gray = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)
plt.imshow(gray, cmap="gray")
plt.title("Grayscale Image")
plt.axis("off")
plt.show()
```

 Convert a given color/gray-scale image into black & white image

```python
# Convert to black & white (thresholding)
_, bw = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)
plt.imshow(bw, cmap="gray")
plt.title("Black & White Image")
plt.axis("off")
plt.show()
```

 Draw the image profile.

```python
# Draw the image profile (intensity values of row 100)
row = gray[100, :]   # take row 100
plt.plot(row)
plt.title("Image Profile (Row 100 Intensity)")
plt.xlabel("Column Index")
plt.ylabel("Intensity")
plt.show()
```

Separate a given color image into three R G & B planes.

```python
# Separate into R, G, B planes
R = img[:,:,0]
G = img[:,:,1]
B = img[:,:,2]

plt.figure(figsize=(12,4))
plt.subplot(1,3,1); plt.imshow(R, cmap="Reds"); plt.title("Red Plane"); plt.axis("off")
plt.subplot(1,3,2); plt.imshow(G, cmap="Greens"); plt.title("Green Plane"); plt.axis("off")
plt.subplot(1,3,3); plt.imshow(B, cmap="Blues"); plt.title("Blue Plane"); plt.axis("off")
plt.show()
```

Create a color image using separate three R, G and B planes.

```python
# Merge R, G, B planes back
merged = cv2.merge([R, G, B])
plt.imshow(merged)
plt.title("Merged Image")
plt.axis("off")
plt.show()
```

```python
# Write given 2D data to an image file
data_2d = np.array([[0, 128, 255],
                    [64, 192, 128],
                    [255, 0, 64]], dtype=np.uint8)
cv2.imwrite("output_data_image.png", data_2d) # Save the 2D array as an image file
img_using_skimage_io = io.imread("output_data_image.png")
# plt.imshow(img_using_skimage_io,cmap="gray") # gray scale image reading
plt.imshow(img_using_skimage_io) #Reading an Image from file written from 2D data 
plt.axis("off")
plt.show()
```
