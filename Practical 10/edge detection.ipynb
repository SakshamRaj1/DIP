### **10. Write and execute a program for edge detection using different edge detection mask.**

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from skimage import data
```

```python
img = data.camera()
plt.imshow(img, cmap='gray')
```

Sobel Edge Detection

```python
# Gradient X
sobelx = cv2.Sobel(img, cv2.CV_64F, 1, 0, ksize=3)
# Gradient Y
sobely = cv2.Sobel(img, cv2.CV_64F, 0, 1, ksize=3)
# Combined Magnitude
sobel_combined = cv2.magnitude(sobelx, sobely)
plt.imshow(sobel_combined, cmap='gray')
```

Laplacian Edge Detection

```python
laplacian = cv2.Laplacian(img, cv2.CV_64F)
plt.imshow(laplacian, cmap='gray')
```

Canny Edge Detection (Most robust)

```python
canny = cv2.Canny(img, 100, 200)
plt.imshow(canny, cmap='gray')
```

Roberts

```python
roberts_x = np.array([[1, 0],
                      [0, -1]])
roberts_y = np.array([[0, 1],
                      [-1, 0]])
edge_roberts_x = cv2.filter2D(img, -1, roberts_x)
edge_roberts_y = cv2.filter2D(img, -1, roberts_y)
edges_roberts = np.sqrt(edge_roberts_x**2 + edge_roberts_y**2)
```

Prewitt

```python
prewitt_x = np.array([[-1, 0, 1],
                      [-1, 0, 1],
                      [-1, 0, 1]])
prewitt_y = np.array([[1, 1, 1],
                      [0, 0, 0],
                      [-1, -1, -1]])
edges_prewitt = np.sqrt(cv2.filter2D(img, -1, prewitt_x)**2 +
                        cv2.filter2D(img, -1, prewitt_y)**2)
```
