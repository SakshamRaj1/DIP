```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from skimage import data, filters
```

```python
image = data.camera()   # already grayscale
image = (image * 1).astype(np.uint8)  # ensure uint8 format
```

```python
kernel_point = np.array([[-1, -1, -1],
                         [-1,  8, -1],
                         [-1, -1, -1]])

points = cv2.filter2D(image, -1, kernel_point)
plt.imshow(points, cmap='gray')
plt.title("Detection of Isolated Points")
plt.axis('off')
```
