```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from skimage import data, filters
```

```python
image = data.camera()   # already grayscale
image = (image * 1).astype(np.uint8)  # ensure uint8 format
```

```python
kernel_horizontal = np.array([[-1, -1, -1],
                              [ 2,  2,  2],
                              [-1, -1, -1]])

kernel_vertical = np.array([[-1, 2, -1],
                            [-1, 2, -1],
                            [-1, 2, -1]])

kernel_45 = np.array([[2, -1, -1],
                      [-1, 2, -1],
                      [-1, -1, 2]])

kernel_135 = np.array([[-1, -1, 2],
                       [-1, 2, -1],
                       [2, -1, -1]])

line_h = cv2.filter2D(image, -1, kernel_horizontal)
line_v = cv2.filter2D(image, -1, kernel_vertical)
line_45 = cv2.filter2D(image, -1, kernel_45)
line_135 = cv2.filter2D(image, -1, kernel_135)
```

```python
plt.imshow(line_h, cmap='gray')
plt.title("Line Detection (Horizontal)")
plt.axis('off')
```

```python
plt.imshow(line_v, cmap='gray')
plt.title("Line Detection (Vertical)")
```

```python
plt.imshow(line_45, cmap='gray')
plt.title("Line Detection (45 Degrees)")
```

```python
plt.imshow(line_135, cmap='gray')
plt.title("Line Detection (135 Degrees)")
```
