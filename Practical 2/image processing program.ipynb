```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from skimage import data
```

```python
# Load Image (Using camera for grayscale operations)
img = data.camera()
plt.imshow(img, cmap='gray')
```

Image Negative

```python
# L - 1 - r (Where L=256)
img_negative = 255 - img
plt.imshow(img_negative, cmap='gray')
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read the image
img = data.camera()

# Obtain negative image
negative = 255 - img

# Display images
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(img, cmap='gray')
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(negative, cmap='gray')
plt.title('Negative Image')
plt.axis('off')

plt.show()
```

Obtain Flip Image

```python
# Horizontal flip
img_flip_horizontal = cv2.flip(img, 1)

# Vertical flip
img_flip_vertical = cv2.flip(img, 0)

# Both horizontal and vertical flip
img_flip_both = cv2.flip(img, -1)

# Display results
plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.imshow(img, cmap='gray')
plt.title('Original Image')
plt.axis('off')

plt.subplot(2, 2, 2)
plt.imshow(img_flip_horizontal, cmap='gray')
plt.title('Horizontal Flip')
plt.axis('off')

plt.subplot(2, 2, 3)
plt.imshow(img_flip_vertical, cmap='gray')
plt.title('Vertical Flip')
plt.axis('off')

plt.subplot(2, 2, 4)
plt.imshow(img_flip_both, cmap='gray')
plt.title('Both Axis Flip')
plt.axis('off')

plt.tight_layout()
plt.show()
```

Thresholding

```python
# Set threshold value
T = 127

# Apply thresholding
_, img_threshold = cv2.threshold(
    img, T, 255, cv2.THRESH_BINARY
)

# Display images
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(img, cmap='gray')
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(img_threshold, cmap='gray')
plt.title('Thresholded Image')
plt.axis('off')

plt.show()
```

Contrast Stretching (Simple Min-Max Normalization)

```python
min_val = np.min(img)
max_val = np.max(img)
img_stretch = ((img - min_val) / (max_val - min_val)) * 255
img_stretch = np.array(img_stretch, dtype=np.uint8)
plt.imshow(img_stretch, cmap='gray')
```
