### **3. To write and execute programs for image arithmetic operations:**

a. Addition of two images

b. Subtract one image from other image

c. Calculate mean value of image

d. Different Brightness by changing mean value

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from skimage import data, img_as_ubyte
from skimage.transform import resize
```

```python
img1 = data.camera() # Grayscale 512x512
    # Create a second image for arithmetic (e.g., a gradient)
img2 = np.zeros((512, 512), dtype=np.uint8)
for i in range(512):
    img2[:, i] = i // 2
```

```python
plt.imshow(img1, cmap='gray')
```

```python
plt.imshow(img2, cmap='gray')
```

Addition of two images

```python
# Using cv2.add handles saturation (caps at 255), numpy adds modulo 256
img_add = cv2.add(img1, img2)
plt.imshow(img_add, cmap='gray')
```

Subtract one image from other

```python
img_sub = cv2.subtract(img1, img2)
plt.imshow(img_sub, cmap='gray')
```

Calculate mean value of image

```python
mean_val = np.mean(img1)
print(f"Mean value of Camera image: {mean_val:.2f}")
```

Different Brightness by changing mean value

```python
brightness_factor = 150
# Increase brightness
img_bright = cv2.add(img1, brightness_factor) 
plt.imshow(img_bright, cmap='gray')
```
