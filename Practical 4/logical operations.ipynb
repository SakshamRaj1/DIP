## **4. To write and execute programs for image logical operations:**

a. AND operation between two images

b.OR operation between two images

c. Calculate intersection of two images

d. Water Marking using X-OR operation

e. NOT operation (Negative image)

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from skimage import data
```

```python
# Load Image
img = data.camera()
```

```python
# Create Binary Image 1 (Thresholded Camera)
_, bin_img1 = cv2.threshold(img, 128, 255, cv2.THRESH_BINARY)
    
# Create Binary Image 2 (Synthetic Circle Mask)
bin_img2 = np.zeros((512, 512), dtype=np.uint8)
cv2.circle(bin_img2, (256, 256), 150, 255, -1)
```

AND operation

```python
img_and = cv2.bitwise_and(bin_img1, bin_img2)
plt.imshow(img_and, cmap='gray')
```

OR operation

```python
img_or = cv2.bitwise_or(bin_img1, bin_img2)
plt.imshow(img_or, cmap='gray')
```

Intersection

```python
# (Mathematically equivalent to AND for binary sets)
img_intersect = img_and
plt.imshow(img_intersect, cmap='gray')
```

NOT operation (Negative)

```python
img_not = cv2.bitwise_not(bin_img1)
plt.imshow(img_not, cmap='gray')
```

XOR operation

```python
# XORing the image with a mask, then XORing again recovers it.
# Here we show the result of XORing the camera image with the circle mask.
img_xor = cv2.bitwise_xor(img, bin_img2)
plt.imshow(img_xor, cmap='gray')
```
