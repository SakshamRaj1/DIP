### **6. To write and execute a program performing the following geometric transformations on an image:**
a. Translation

b. Scaling

c. Rotation

d. Shrinking

e. Zooming

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from skimage import data
```

```python
img = data.camera()
rows, cols = img.shape
plt.imshow(img, cmap='gray')
```

Translation

```python
# Shift by (100, 50)
M_trans = np.float32([[1, 0, 100], [0, 1, 50]])
img_trans = cv2.warpAffine(img, M_trans, (cols, rows))
plt.imshow(img_trans, cmap='gray')
```

Scaling & 

e. Zooming (Scaling Up)

```python
# Scale by 2x
img_zoom = cv2.resize(img, None, fx=2, fy=2, interpolation=cv2.INTER_CUBIC)
plt.imshow(img_zoom, cmap='gray')
```

Shrinking (Scaling Down)

```python
# Scale by 0.5x
img_shrink = cv2.resize(img, None, fx=0.5, fy=0.5, interpolation=cv2.INTER_AREA)
plt.imshow(img_shrink, cmap='gray')
```

Rotation

```python
# Rotate 45 degrees around center
center = (cols/2, rows/2)
M_rot = cv2.getRotationMatrix2D(center, 45, 1)
img_rot = cv2.warpAffine(img, M_rot, (cols, rows))
plt.imshow(img_rot, cmap='gray')
```
