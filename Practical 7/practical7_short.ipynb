### To understand various image noise models and to write programs for:

a. Image restoration

b. Remove Salt and Pepper Noise

c. Minimize Gaussian noise

d. Median filter and Weiner filter

```python
import matplotlib.pyplot as plt
from skimage import data
from skimage.util import random_noise
from scipy.ndimage import median_filter, gaussian_filter
from scipy.signal import wiener
```

```python
# Load camera image
img = data.camera()

# Noise models
sp_noisy = random_noise(img, mode='s&p', amount=0.05)
gauss_noisy = random_noise(img, mode='gaussian', var=0.01)

# Restoration filters
sp_median = median_filter(sp_noisy, size=3)       # Removes Salt & Pepper
gauss_blur = gaussian_filter(gauss_noisy, sigma=1) # Minimizes Gaussian noise
gauss_wiener = wiener(gauss_noisy, (5, 5))         # Wiener restoration

# Display results
titles = ['Original', 'Salt & Pepper Noise', 'Median Filter',
          'Gaussian Noise', 'Gaussian Filter', 'Wiener Filter']
images = [img, sp_noisy, sp_median, gauss_noisy, gauss_blur, gauss_wiener]

plt.figure(figsize=(12, 7))
for i, (im, title) in enumerate(zip(images, titles), 1):
    plt.subplot(2, 3, i)
    plt.imshow(im, cmap='gray')
    plt.title(title)
    plt.axis('off')
plt.tight_layout()
plt.show()
```
